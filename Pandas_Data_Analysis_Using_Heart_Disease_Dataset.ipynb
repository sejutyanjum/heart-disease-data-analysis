#Categorical Data Analysis

```python
#Count the number of patients with heart disease (target = 1) and without heart disease (target = 0)
df['target'].value_counts()
```

```python
#Count the number of male (sex = 1) and female (sex = 0) patients
df['sex'].value_counts()
```

#Data Selection Using iloc

```python
#Select the first 10 rows of the dataset
df.iloc[0:10]
```

```python
#Extract the columns representing age, sex, and cholesterol
df.iloc[0:,[0,1,4]]
```

```python
#Select rows from index 20 to 30 and columns from index 0 to 4
df.iloc[20:31,0:5]
```

#Data Filtering Using Conditions

```python
#Extract all patients older than 50 years
df[df['age']>50]
```

```python
#Extract patients who have heart disease and a cholesterol level greater than 240
df[(df['target']==1) & (df['chol']>240)]
```

```python
#Count how many patients satisfy both conditions
((df['target']==1) & (df['chol']>240)).sum()
```
